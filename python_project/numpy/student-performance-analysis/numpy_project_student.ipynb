```python
# Student Performance Analysis System
```

```python
import numpy as np
```

```python
# marks of student 
marks_of_std = np.random.randint(35,101,(100,5))
print(marks_of_std)
```

```python
# total marks of each strudent

print('marks of the each student:')
total_marks = np.sum(marks_of_std, axis =1)

total_result = np.vstack(total_marks)
print(total_result)
```

```python
# 3. Calculate Average Marks(student wise and subject wise)

average_marks_std = np.average(marks_of_std, axis= 1)
average_marks_sub = np.average(marks_of_std, axis= 0)

average_marks_stdd = np.vstack(average_marks_std)
average_marks_subb = np.vstack(average_marks_sub)

print('student wise average:')
print(average_marks_stdd)
print('subject wise average:')
print(average_marks_subb)
```

```python
# 4. Find Topper

topper = np.max(total_result)
topper_index = np.argmax(total_result)
print(topper)
print(topper_index)
```

```python
# 5. Find Failed and distinction Students

print('failed_student:', marks_of_std[marks_of_std < 40])
print('')

print('distinction_student:')
dist =  marks_of_std[marks_of_std > 80]
print(np.vstack(dist))
```

```python
# 6. Grade System

empty_arr = np.empty(marks_of_std.shape, dtype= object)

empty_arr[marks_of_std >= 85] = 'A' 

empty_arr[(marks_of_std >= 70) & (marks_of_std < 85) ] = 'B'

empty_arr[(marks_of_std >= 40) & (marks_of_std < 70) ] = 'C'

empty_arr[marks_of_std < 40] = 'Fail'

empty_array = np.hstack(empty_arr)
print(empty_arr)

```

```python
# 7. Subject Analysis

# average of all subject
avgOfSub = np.mean(marks_of_std, axis=0)

print(avgOfSub)

# hardest subject 
hardest_sub = np.argmin(avgOfSub)

# easiest subject
easiest_sub = np.argmax(avgOfSub)

print(hardest_sub)
print(easiest_sub)

print(avgOfSub[easiest_sub])
```

```python
# 8. Rank Students

rank = np.argsort(total_marks)

print(rank)
print('')
print(total_marks[rank])
```

```python
# Attendance Dataset

attendence = np.random.randint(1,100, 100)

attend=np.vstack(attendence)
print(attend)
```

```python
# 10. students below 75%

print('below attendence index:', np.argsort(attend[attend < 75]))
print('')
print('below attendence:', attend[attend < 75])
```

```python
# Check student-wise failure
failed_students = np.any(empty_arr == 'Fail', axis=1)

# Count failed students
failed_count = np.sum(failed_students)

# Count passed students
passed_count = len(empty_arr) - failed_count

print("Passed Students:", passed_count)
print("Failed Students:", failed_count)
```

```python
# percentage of how many students get fail or pass
labels = ["Pass", "Fail"]
values = [passed_count,failed_count]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.show()
```
